```python
#clean
df = df.dropna()
```

```python
df = df.drop_duplicates()
df
```

```python
#Avg
avg_open = df['Open'].mean()
avg_close = df['Close'].mean()

print(f"Average Open Price: {avg_open:.2f}")
print(f"Average Close Price: {avg_close:.2f}")
```

```python
highestinhigh = df['High'].max()
lowestinlow = df['Low'].min()

print(f"Highest High Price: {highestinhigh:.2f}")
print(f"Lowest Low Price: {lowestinlow:.2f}")
```

```python
df['Delta'] = df['Close'] - df['Open']
print(df[['Open', 'Close', 'Delta']])
```
